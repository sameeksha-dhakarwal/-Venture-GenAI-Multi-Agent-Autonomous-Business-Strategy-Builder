```python
labels = ["Pass", "Fail"]
values = [100, 0]

plt.figure()

# Create donut
wedges, texts, autotexts = plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

# Draw center circle
centre_circle = plt.Circle((0, 0), 0.70)
plt.gca().add_artist(centre_circle)

plt.title("System Success Rate (Donut Style)")
plt.show()
```

```python
features = ["Accuracy", "Relevance", "Hallucination"]

with_rag = [5, 5, 1]
without_rag = [3, 2, 5]

x = np.arange(len(features))

plt.figure()
plt.bar(x - 0.2, with_rag, width=0.4, label="With RAG")
plt.bar(x + 0.2, without_rag, width=0.4, label="Without RAG")

plt.xticks(x, features)
plt.legend()
plt.title("Performance Comparison (RAG vs No RAG)")
plt.show()
```

```python
agents = ["Market", "Competitor", "Business", "Finance", "Pitch"]

relevance = [4.6, 4.3, 4.7, 4.1, 4.8]
completeness = [4.5, 4.2, 4.6, 4.6, 4.7]
logic = [4.4, 4.1, 4.5, 4.2, 4.6]

x = np.arange(len(agents))

plt.figure()
plt.bar(x - 0.25, relevance, width=0.25, label="Relevance")
plt.bar(x, completeness, width=0.25, label="Completeness")
plt.bar(x + 0.25, logic, width=0.25, label="Logical Consistency")

plt.xticks(x, agents)
plt.legend()
plt.title("Agent-wise Metric Comparison")
plt.ylabel("Score (out of 5)")
plt.show()
```

```python
cm = np.array([[90, 5],
               [3, 92]])

plt.figure()
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
TP, FP, FN, TN = 90, 5, 3, 92

accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)

print("Accuracy:", round(accuracy, 3))
print("Precision:", round(precision, 3))

plt.figure()
plt.bar(["Accuracy", "Precision"], [accuracy, precision])
plt.ylim(0,1)
plt.title("Accuracy vs Precision")
plt.show()
```
